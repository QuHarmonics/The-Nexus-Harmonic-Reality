## Lead-in

Put some indistinguishable tokens on the vertices of a finite graph, one per vertex, and let a
token step to an empty neighbour. That is the whole system. The graph is the *substrate*; the space
of configurations and legal steps is the *arena*.

Now change something about it — add an edge, subdivide one, glue two vertices together, allow a
site to hold two tokens instead of one, or simply look at the system with a coarser symmetry. Each
of those changes alters which transformations exist. The question this paper asks is narrow and
answerable:

> **When the rules of movement change, what happens to the structure that was there before?**

The naive answers are that structure is created or destroyed. Both turn out to be wrong in a
precise and checkable sense. Every elementary change either *extends* the transformation structure,
*re-expresses* it, or *hides* it from a particular observer. Structure is never annihilated. The
one operation that does produce a genuine kernel — restricting which configurations are admissible
— produces it as a **relative cohomology class**: something present in the larger admissible space
and absent from the smaller one, with an exact dimension formula.

The headline object is a single integer attached to each such restriction. For a distinguished site
allowed to hold $r$ tokens instead of $r+1$, that integer is the first Betti number of a small
quotient graph, and it can be computed without ever building the arena.

A concrete instance, which the reader can verify in Section 9: on the 7-cycle with four tokens, the
number of independent transformation channels visible to a symmetric observer goes
$16 \to 27 \to 31 \to 31$ as the distinguished site's capacity is relaxed from one token to four.
The increments $11, 4, 0$ are the capacity defects. The final zero is **not** a saturation effect —
it is a statement about the local symmetry at that site, and on other substrates it is nonzero.

## 1. Setup

$X = (V,E)$ is a finite connected simple graph, $K \ge 1$ a token count, $G \le \operatorname{Aut}(X)$.

**Arena.** $\mathcal A(X,K)$ has as vertices the $K$-subsets $S \subseteq V$ and as edges the moves
$S \to S - u + w$ with $uw \in E$, $u \in S$, $w \notin S$.

**Lemma 1.1 (edge parametrisation).** Arena edges are in $G$-equivariant bijection with pairs
$(e,T)$, $e = \{u,w\} \in E$, $T \subseteq V \setminus e$, $|T| = K-1$. The two orientations are
the two choices of which endpoint of $e$ carries the token at the source.

*Proof.* $T = S \cap S'$ and $e$ is the moved token's base edge; the correspondence is manifestly
equivariant. $\square$

Everything downstream follows from this. It is the reason the base graph, not the arena, controls
most of the answers.

**Residual.** With $C^0, C^1$ the real cochain groups of the arena 1-skeleton ($C^1$ antisymmetric
under edge reversal),

$$R_G(X,K) \;=\; \frac{(C^1)^G}{d_0\big((C^0)^G\big)}, \qquad \epsilon(X,K;G) = \dim R_G(X,K).$$

Averaging over the finite group gives $R_G \cong H^1(\mathcal A)^G$ over $\mathbb R$; the quotient
presentation is the one used below. Three distinct objects appear in the literature on this and
must not be conflated: $H^1(\mathcal A;\mathbb R)$, its invariant part $R_G$, and
$H^1$ of the complex with commuting squares filled. Every statement here names which one it means.

## 2. What $\epsilon$ is

**Theorem 2.1.** $\dim (C^1)^G = e_G^{\rm ns}$, the number of $G$-orbits of arena edges whose
stabiliser contains no orientation-reversing element.

*Proof.* Orbit–stabiliser on oriented edges. If some $g$ reverses $e$ then invariance forces
$\omega(e) = -\omega(e)$, hence $\omega(e)=0$; otherwise one scalar determines $\omega$ on the whole
orbit. $\square$

**Theorem 2.2.** For connected $\mathcal A$, $\dim d_0((C^0)^G) = v_G - 1$, $v_G$ the number of
configuration orbits. Hence

$$\boxed{\ \epsilon = e_G^{\rm ns} - v_G + 1 = \beta_1\big(\mathcal A /\!/ G\big)\ }$$

where $\mathcal A/\!/G$ is the quotient graph with the reversible edge orbits deleted.

The three descriptions — orbit counting, cochain dimensions, quotient-graph topology — are the same
integer. Measured across 30 arena/predicate combinations with no discrepancy.

**Theorem 2.3 (closed form).** Writing $P_g(x) = \prod_{c \in \operatorname{Cyc}(g)}(1+x^{|c|})$,
$m^+(g)$ for edges of $X$ fixed pointwise by $g$ and $m^-(g)$ for edges whose endpoints $g$ swaps,

$$\epsilon(X,K;G) = 1 + \frac{1}{|G|}\sum_{g \in G}\left(m^+(g)\,[x^{K-1}]\tfrac{P_g}{(1+x)^2} - m^-(g)\,[x^{K-1}]\tfrac{P_g}{1+x^2} - [x^K]P_g\right)$$

Cycle structure of $G$ on $V(X)$ plus two edge counts per group element. No arena is built. Verified
against direct orbit counting on 106 (substrate, $K$, block-profile) cases across five base graphs
including a non-vertex-transitive one, $K = 2..5$: no discrepancies.

Two immediate consequences. $\epsilon(K_N,K) = 0$ for all $N,K$ under the full symmetric group —
$S_N$ is transitive on $(e,T)$ and the transposition $(u\,w)$ reverses $e$ while fixing $T$
pointwise, so the single edge orbit is reversible. And for the trivial group there is no quotient at
all, so $\epsilon(X,K;\{1\}) = \beta_1(\mathcal A(X,K))$ exactly — not a limit.

**Measured, not proved: $\mathcal H^1_G = 0$.** Across three arenas and both standard 2-cell
predicates (disjoint-support commuting squares; all arena 4-cycles), the invariant harmonic space
was zero in every case, so the entire invariant non-exact sector is coexact. Cell counts moved by a
factor of 5.7 without moving $\epsilon$ by one dimension: the residual is a property of the
1-skeleton and the group, and the 2-cells only decide how it splits. No case with
$\mathcal H^1_G \neq 0$ is known.

## 3. What a reader can see

A *reader* coarse-grains the substrate into blocks and records block-to-block flow. Two natural
responses exist, and both are blind to the residual, for different reasons.

The scalar response $Q_\ell = d\psi_\ell$, $\psi_\ell(S) = \sum_B |B \cap S|^2$, is exact, so its
integral round any loop vanishes — in the arena and in any quotient of it.

The matrix response is subtler. Contracting the antisymmetric block-flow table with any fixed
antisymmetric $M$ gives $2M_{b(u)b(w)}$ for a move $u \to w$: it depends only on the moved **base
edge and its orientation**. Hence every matrix-reader cochain is a pullback $\pi^*\eta$ along the
equivariant map $\pi$ from arena directed edges to base directed edges.

**Theorem 3.1 (reader bound).** $\dim \operatorname{proj}_R(\pi^*C^1(X)) \le \epsilon(X,1;G)$.

*Proof.* Equivariance gives $\langle r, \pi^*\eta\rangle = \langle r, \pi^*(\Pi_X\eta)\rangle$, so
only invariant base forms contribute. And $\pi^*(d_X f) = d_{\mathcal A}F$ with
$F(S) = \sum_{v \in S} f(v)$, so exact base forms pull back to coboundaries. What remains is the
base graph's own residual at one token, and $\mathcal A(X,1) = X$. $\square$

Measured: the bound is attained and is independent of $K$. Nineteen cases; on every symmetric
substrate the reach is $0$ — $0$ of $2$ channels on $C_8$ at $K=3$, $0$ of $3$ at $K=4$, $0$ of $1$
on $Q_3$ and Petersen — and on deliberately constructed trivial-automorphism bases with
$\epsilon(X,1) = 1, 2$ the reach is exactly $1,1,1$ and $2,2,2$ across $K = 2,3,4$ against arena
residuals of $10,17,10$ and $14,23,14$.

$$\boxed{\ \epsilon(X,K) - \epsilon(X,1) = \text{channels created by } K \text{ tokens sharing a substrate, unreachable by any one-token observable}\ }$$

## 4. The four motions

Let $X \to X'$ be an elementary change. The residual is only defined relative to a group, and the
group changes too, so comparisons pass through $G_\cap = \operatorname{Aut}(X) \cap
\operatorname{Aut}(X')$. There is no map $R_X \to R_{X'}$; there is a span.

| motion | what happens to a transformation | induced map on residuals | measured |
|---|---|---|---|
| **extension** — add a base edge | survives verbatim; new ones appear | $\Delta_R$ onto, $\operatorname{coker}=0$ | 6 pairs |
| **re-factorisation** — subdivide an edge | $[u_0 \to w_0] \mapsto [u_0 \to z] + [z \to w_0]$ | $F^*_R$ onto, $\operatorname{coker}=0$ | 5 pairs, $\partial F = F\partial$ to $0.0$ |
| **hiding** — enlarge the group | unchanged | $R_{G'} \hookrightarrow R_G$, $\ker = 0$ | 5 pairs |
| **identification** — contract an edge | configurations merge, $q$ many-to-one | $q^*$ injective; kernel only via admissibility | Section 5 |

No motion removes a channel. The one case that looks like removal — $P_6 \to C_6$ takes the residual
from 5 to 0 — added six transformations and destroyed none: the five channels persist as
$\mathbb Z_2$-invariant cochains, and $R_{D_6} = 0$ is a statement about the enlarged observer.

The same correction applies to a number that motivated much of this work. $\epsilon(C_N,3) = 0,1,2,3$
for $N = 6,7,8,9$ looks like loops appearing as the cycle lengthens. Pulling the $C_7$ channel back
along the subdivision chain map gives rank $1$ of $1$, landing inside $R_{C_6}$ at the common group;
likewise $2$ of $2$ and $3$ of $3$. The cochains are present throughout. The shorter cycle's
dihedral group cannot see them.

$$\boxed{\ \text{edits extend or re-factor; symmetry hides; projection erases visibility, not structure}\ }$$

## 5. Admissibility: the capacity filtration

Fix a distinguished site $m \in V$ and let $G = \operatorname{Stab}_{\operatorname{Aut}(X)}(m)$. Let
$X^{(r)}$ be the arena in which $m$ may hold up to $r$ tokens and every other site at most one.
Then $X^{(r)} \subseteq X^{(r+1)}$, with restriction
$\sigma_r : R_G(X^{(r+1)}) \to R_G(X^{(r)})$ and newly admitted layer

$$\mathcal D_r = \{\, m^{r+1} + U \;:\; U \subseteq V\setminus\{m\},\ |U| = K-r-1 \,\}.$$

**Theorem 5.1.** $\sigma_r$ is surjective; $\operatorname{coker}\sigma_r = 0$.

*Proof.* Extend an invariant 1-cochain by $0$ on the new edges and average; the extension is
invariant and restricts correctly. For a graph $C^2 = 0$, so every 1-cochain is a cocycle and the
extension descends. $\square$

**Definition.** $D_r(G,X,K) = \ker\sigma_r$, the **capacity defect**, and $p_r = \dim D_r$.

**Theorem 5.2.** $D_r \cong H^1_G\big(X^{(r+1)}, X^{(r)}\big)$.

*Proof.* If $[\omega] \in \ker\sigma_r$ then $\omega|_{X^{(r)}} = d_0 f$ with $f$ invariant; extend
$f$ and average to $\tilde f$, which restricts to $f$ because $G$ preserves the layers. Then
$\omega - d_0\tilde f$ vanishes on $X^{(r)}$, so every kernel class has a representative supported
on the new layer. The converse is immediate. $\square$

### 5.1 The cone theorem

The relative complex is the reduced complex of the quotient space: $X^{(r)}$ is **collapsed to a
single point $*$**, not deleted. Define the graph $\mathcal Q_r$:

- vertices: the $G$-orbits of $\mathcal D_r$, **plus the cone point $*$**;
- edges: the $G$-orbits of arena moves with at least one endpoint in $\mathcal D_r$ (cross moves
  land on $*$; internal moves join two layer orbits), with the reversible orbits deleted.

**Theorem 5.3 (cone).** $\dim D_r = \beta_1(\mathcal Q_r) = |E_Q| - |V_Q| + c_Q$.

*Proof.* By Theorem 2.1 applied relatively, $\dim C^1_G = |E_Q|$. Relative 0-cochains are functions
on the non-cone vertices, so $\dim C^0_G = |V_Q| - 1$. The relative coboundary is the ordinary graph
coboundary with the value at $*$ pinned to zero, so its kernel is the constants on each component
with the $*$-component forced to zero: $\dim\ker = c_Q - 1$, hence
$\operatorname{rank} = |V_Q| - c_Q$. Since $C^2 = 0$,
$\dim D_r = |E_Q| - (|V_Q| - c_Q)$. $\square$

Since $|V_Q| = v_{G,\mathcal D_r} + 1$ and $|E_Q| = e^{ns}_{G,\mathcal D_r}$:

$$\dim D_r = e^{ns}_{G,\mathcal D_r} - v_{G,\mathcal D_r} - 1 + c_Q, \qquad\text{and}\qquad D_r = 0 \iff \mathcal Q_r \text{ is a forest}.$$

The cone point is not bookkeeping. Without it the saturated case would return $\beta_1 = -1$.

### 5.2 Connectivity is automatic

**Lemma 5.4 (reversible orbits are loops).** If $g \in G$ reverses the arena edge $\{S,T\}$ then
$gS = T$, so $[S]_G = [T]_G$ and the orbit is a self-loop in the quotient. Deleting self-loops
cannot change components, so $c(\mathcal Q_r) = c(\widehat{\mathcal Q}_r)$, the count before
deletion. $\square$

**Lemma 5.5 (vacancy propagation).** Let $S = m^{r+1}+U$ with a vacancy $v \notin \{m\}\cup U$. If
$N(m) \subseteq U$ there is no immediate move to $X^{(r)}$, but along a substrate path from $v$ to
$N(m)$ the first occupied-to-vacant edge $x \to y$ has $x \neq m$, so the move preserves the
occupancy $r+1$ and stays inside $\mathcal D_r$. Iterating frees a neighbour of $m$; then
$m \to v_j$ lands in $X^{(r)}$. $\square$

$$\boxed{\ N(m) \subseteq U \ \not\Rightarrow\ c(\mathcal Q_r) > 1 \ }$$

**Proposition 5.6 (saturation is degenerate).** $c(\mathcal Q_r) > 1$ requires
$U = V\setminus\{m\}$, i.e. $K = |V| + r$. But a configuration of $X^{(r)}$ needs
$|U| = K - c \le |V| - 1$ with $c \le r$, forcing $c \ge K - |V| + 1 = r+1$, a contradiction.
Hence $K = |V|+r$ implies $X^{(r)} = \varnothing$. $\square$

So whenever the old arena is nonempty, $c(\mathcal Q_r) = 1$ and

$$\boxed{\ \dim D_r = e^{ns}_{G,\mathcal D_r} - v_{G,\mathcal D_r}\ }$$

An exhaustive search over 14 substrates, every site orbit, $K = 2\ldots|V|+2$, $r = 1\ldots K$ —
**492 capacity transitions** — found: $\operatorname{coker}\sigma_r = 0$ in all; $\dim D_r =
\beta_1(\mathcal Q_r)$ in all; reversible orbits self-loops in all; and **zero** cases of $c > 1$
with $X^{(r)}$ nonempty. The 48 cases with $c = 2$ all have $X^{(r)} = \varnothing$, which is the
degenerate corner of Proposition 5.6.

This corrects a reading that circulated during the derivation: the saturated $c=2$ example is not an
"admissible but unreachable sector," because there is nothing for it to be unreachable *from*. No
non-degenerate disconnected layer is known.

### 5.3 The top layer

At $r = K-1$ the layer is the single configuration $m^K$, so $\mathcal Q_{K-1}$ has two vertices and
one edge per $\operatorname{Stab}(m)$-orbit on $N(m)$ — none reversible, since the endpoints are
different configurations. Hence

$$\boxed{\ \dim D_{K-1} = \#\big(N(m)/\operatorname{Stab}(m)\big) - 1\ }$$

zero exactly when the stabiliser is transitive on the neighbours of $m$. Verified 47/47.

This kills the reading that the ladder $11,4,0$ terminates because capacity saturates. Every
substrate in that ladder happened to have a locally transitive stabiliser. The prism at $m$ of
degree 3 with $|\operatorname{Stab}| = 2$ gives $D_{K-1} = 1$; an interior path vertex with trivial
stabiliser gives 1; the paw graph gives 2. **Local symmetry controls the terminal value; global
saturation controls only the degenerate disconnection.** Different mechanisms.

## 6. Three levels

Forgetting the group gives the field-side defect $D_r^{\rm ord} = H^1(X^{(r+1)},X^{(r)})$; the
observer-side defect is its invariant part. The difference

$$L = \dim D_r^{\rm ord} - \dim D_r$$

is the sector that exists as transformation cycles but cannot be represented invariantly by this
observer. For $C_7$ at $K=4$: $(25,9,1)$ ordinary against $(11,4,0)$ invariant, so $L = (14,5,1)$.
For $Q_3$ at $K=4$: $(69,20,2)$ against $(11,3,0)$, $L = (58,17,2)$.

$$\text{admissibility } X^{(r)} \subset X^{(r+1)} \;\longrightarrow\; \text{field-relative sector} \;\longrightarrow\; \text{observer-visible sector}$$

Downward, $\sigma_r$ looks like loss. Upward, the same object is newly exposed relative structure.
The mathematics establishes the filtering; naming $L$ "information loss" would require an
independently defined observation map and is not done here.

## 8. Corrections made during the derivation

These are recorded because each one is a distinction that the final statement depends on.

1. *"$\epsilon$ measures creation of relational structure."* False. Subdivision pullback is full
   rank on the cycle family: the channels pre-exist and the shorter cycle's group cannot see them.
2. *"$\epsilon(X,2) = 0$ universally."* False — it held only for vertex-transitive members.
   $\epsilon(P_8,2) = 6$, $\epsilon(P_6,2) = 2$.
3. *"$\epsilon = $ chiral-pair count."* Matched at $K=3$ for every $N$ and fails from $K=4$:
   $\epsilon(C_{12},\cdot) = 9,32,63,77$ against chiral $9,14,28,30$.
4. *"Graded $\iff m \ge 3$ blocks and max block $\ge K$."* Fitted 28/28 at $K=3$, broken at $K=4$.
5. *"The three-block threshold."* Wrong side: the threshold is $K \ge 3$ tokens, a property of the
   system, not the reader.
6. *"Grading requires a multi-token transition."* False; single-token moves grade.
7. *"Circulation on $Q_3$ is evidence for the coexact reading."* It was forced by $\epsilon = 1$;
   it confirmed the implementation, nothing more.
8. *"As $G \to 1$, $\epsilon \to \beta_1$."* There is no limit; the trivial group is an exact case.
9. *"The capacity ladder terminates at 0 because capacity saturates."* False — Section 5.3.
10. *"The saturated $c=2$ case is an admissible-but-unreachable sector."* Degenerate: $X^{(r)}$ is
    empty there (Prop. 5.6).
11. A single-reader measurement that appeared to give a length-independent invariant
    $(\sum|Q|,\sum|a|) = (4,8)$ on both $C_6$ and $C_8$ did not survive sweeping all readers.
12. In preparing this paper, a test of the top-layer formula reported 16 failures out of 47. The
    failure was a destructuring bug in the test harness that forced the measured value to zero. The
    corrected test gives 47/47. Recorded because an uncorrected version of that table would have
    falsified a true theorem.

### 9.1 Library


```python
"""Consolidated verification for the paper. Everything is recomputed from scratch here."""
import itertools, numpy as np
from collections import defaultdict
ATOL=1e-9
def rank_abs(M):
    if M.size==0 or min(M.shape)==0: return 0
    return int((np.linalg.svd(M,compute_uv=False)>ATOL).sum())
def colspace(M):
    if M.size==0 or min(M.shape)==0: return np.zeros((M.shape[0],0))
    U,s,_=np.linalg.svd(M,full_matrices=False); return U[:,s>ATOL]
def autos(E,n):
    Es=set(map(tuple,map(sorted,E))); adj=defaultdict(set); deg=defaultdict(int)
    for u,v in Es: adj[u].add(v); adj[v].add(u); deg[u]+=1; deg[v]+=1
    out,p,used=[],[None]*n,[False]*n
    def bt(i):
        if i==n: out.append(tuple(p)); return
        for c in range(n):
            if used[c] or deg[c]!=deg[i]: continue
            if all((j in adj[i])==(p[j] in adj[c]) for j in range(i)):
                p[i]=c; used[c]=True; bt(i+1); used[c]=False
    bt(0); return out
def cap_arena(adj,n,K,m,r):
    others=[v for v in range(n) if v!=m]; V=[]
    for c in range(0,min(r,K)+1):
        if K-c>len(others): continue
        for U in itertools.combinations(others,K-c): V.append((c,frozenset(U)))
    V=sorted(V,key=lambda s:(s[0],sorted(s[1]))); vi={s:i for i,s in enumerate(V)}
    Es={}
    for (c,U) in V:
        if c>=1:
            for v in adj[m]:
                if v not in U:
                    t=(c-1,frozenset(U|{v}))
                    if t in vi: Es[frozenset(((c,U),t))]=((c,U),t)
        for v in list(U):
            if v in adj[m] and c+1<=r:
                t=(c+1,frozenset(U-{v}))
                if t in vi: Es[frozenset(((c,U),t))]=((c,U),t)
        for u in list(U):
            for w in adj[u]:
                if w==m or w in U: continue
                t=(c,frozenset((U-{u})|{w}))
                if t in vi: Es[frozenset(((c,U),t))]=((c,U),t)
    E=sorted(Es.values(),key=lambda p:(p[0][0],sorted(p[0][1]),p[1][0],sorted(p[1][1])))
    ei={}
    for i,(a,b) in enumerate(E): ei[(a,b)]=(i,1); ei[(b,a)]=(i,-1)
    return V,vi,E,ei
def act(g,s): c,U=s; return (c,frozenset(g[x] for x in U))
def residual(V,vi,E,ei,G):
    if not E: return np.zeros((0,0))
    D0=np.zeros((len(E),len(V)))
    for (a,b) in E:
        i,_=ei[(a,b)]; D0[i,vi[b]]+=1; D0[i,vi[a]]-=1
    P=np.zeros((len(E),)*2)
    for g in G:
        for (a,b) in E:
            i,_=ei[(a,b)]; j,s=ei[(act(g,a),act(g,b))]; P[j,i]+=s/len(G)
    Inv=colspace(P); Ex=colspace(P@colspace(D0))
    return colspace(Inv-Ex@(Ex.T@Inv)) if Ex.shape[1] else Inv
def quotient(adj,n,K,m,r,G):
    V,vi,E,ei=cap_arena(adj,n,K,m,r+1)
    new=lambda s:s[0]==r+1
    rep,seen={},set()
    for s in V:
        if not new(s) or s in seen: continue
        orb={act(g,s) for g in G}
        for t in orb: rep[t]=s
        seen|=orb
    nodes=set(rep.values())|{"*"}; node=lambda s: rep[s] if new(s) else "*"
    edges,seen_e,loops=[],set(),0
    for (a,b) in E:
        if not(new(a) or new(b)): continue
        key=frozenset((a,b))
        if key in seen_e: continue
        seen_e|={frozenset((act(g,a),act(g,b))) for g in G}
        flip=any(act(g,a)==b and act(g,b)==a for g in G)
        if flip:
            loops += (node(a)!=node(b))      # must be 0: flippable orbits are self-loops
            continue
        edges.append((node(a),node(b)))
    par={x:x for x in nodes}
    def f(a):
        while par[a]!=a: par[a]=par[par[a]]; a=par[a]
        return a
    for a,b in edges: par[f(a)]=f(b)
    c=len({f(x) for x in nodes})
    return len(nodes),len(edges),c,len(edges)-len(nodes)+c,loops
def defect(adj,n,K,m,r,G):
    Va,via,Ea,eia=cap_arena(adj,n,K,m,r+1); Vb,vib,Eb,eib=cap_arena(adj,n,K,m,r)
    Ra=residual(Va,via,Ea,eia,G); Rb=residual(Vb,vib,Eb,eib,G)
    if Ra.shape[1]==0: return 0,Rb.shape[1],0,0
    if Rb.shape[1]==0: return Ra.shape[1],0,0,Ra.shape[1]
    rows=[eia[(a,b)][0] for (a,b) in Eb]; sg=np.array([eia[(a,b)][1] for (a,b) in Eb])[:,None]
    rk=rank_abs(Rb.T@(Ra[rows,:]*sg))
    return Ra.shape[1],Rb.shape[1],Rb.shape[1]-rk,Ra.shape[1]-rk

C=lambda q:[(i,(i+1)%q) for i in range(q)]
P=lambda q:[(i,i+1) for i in range(q-1)]
Q3=[(v,v^(1<<i)) for v in range(8) for i in range(3) if v<v^(1<<i)]
PV=sorted(itertools.combinations(range(5),2)); PI={s:i for i,s in enumerate(PV)}
PET=[(PI[s],PI[t]) for s in PV for t in PV if PI[s]<PI[t] and not set(s)&set(t)]
K33=[(i,3+j) for i in range(3) for j in range(3)]
PRISM=[(0,1),(1,2),(2,0),(3,4),(4,5),(5,3),(0,3),(1,4),(2,5)]
TAIL=[(0,1),(1,2),(2,0),(2,3)]           # triangle with a pendant
PAW =[(0,1),(1,2),(2,0),(0,3),(1,4)]
FAM=[("C3",C(3),3),("C4",C(4),4),("C5",C(5),5),("C6",C(6),6),("C7",C(7),7),
     ("P3",P(3),3),("P4",P(4),4),("P5",P(5),5),("P6",P(6),6),
     ("K33",K33,6),("prism",PRISM,6),("Q3",Q3,8),("Petersen",PET,10),
     ("tri+tail",TAIL,4),("paw",PAW,5)]

# ---- epsilon: orbit count and Burnside closed form ----
def edges_of(adj): return sorted({tuple(sorted((u,v))) for u in adj for v in adj[u]})
def brute_aut(adj, n):
    E = set(edges_of(adj)); out = []
    for p in itertools.permutations(range(n)):
        if all(tuple(sorted((p[u], p[v]))) in E for (u, v) in E): out.append(p)
    return out
def dihedral(n):
    return [tuple((i+k) % n for i in range(n)) for k in range(n)] + \
           [tuple((k-i) % n for i in range(n)) for k in range(n)]
def sym(n): return list(itertools.permutations(range(n)))
def aut_pet():
    return [tuple(PET_I[tuple(sorted((p[s[0]], p[s[1]])))] for s in PET_V)
            for p in itertools.permutations(range(5))]
def aut_qd(d):
    V = 1 << d
    return [tuple(sum(((v >> i) & 1) << pm[i] for i in range(d)) ^ m for v in range(V))
            for pm in itertools.permutations(range(d)) for m in range(V)]
def eps_orbits(adj, n, K, G):
    EX = edges_of(adj)
    verts = list(itertools.combinations(range(n), K))
    seen, v_orb = set(), 0
    for S in verts:
        fs = frozenset(S)
        if fs in seen: continue
        seen |= {frozenset(g[i] for i in S) for g in G}; v_orb += 1
    items = [(frozenset(e), frozenset(T))
             for e in EX
             for T in itertools.combinations([x for x in range(n) if x not in e], K-1)]
    seen, e_ns = set(), 0
    for it in items:
        if it in seen: continue
        e, T = it
        orb = {(frozenset(g[i] for i in e), frozenset(g[i] for i in T)) for g in G}
        seen |= orb
        u, w = sorted(e)
        rev = any(g[u] == w and g[w] == u and {g[i] for i in T} == set(T) for g in G)
        if not rev: e_ns += 1
    return e_ns - v_orb + 1, v_orb, e_ns
def poly_mul(a, b):
    r = [0]*(len(a)+len(b)-1)
    for i, x in enumerate(a):
        if x:
            for j, y in enumerate(b): r[i+j] += x*y
    return r
def cycles_of(g):
    n, seen, out = len(g), [False]*len(g), []
    for i in range(n):
        if seen[i]: continue
        L, j = 0, i
        while not seen[j]: seen[j] = True; j = g[j]; L += 1
        out.append(L)
    return out
def P_from(lens):
    p = [1]
    for L in lens: p = poly_mul(p, [1] + [0]*(L-1) + [1])
    return p
def coeff(p, k): return p[k] if 0 <= k < len(p) else 0
def eps_closed(adj, n, K, G):
    EX = edges_of(adj); tot = 0
    for g in G:
        lens = cycles_of(g)
        P = P_from(lens)
        Fv = coeff(P, K)
        mplus = sum(1 for (u, v) in EX if g[u] == u and g[v] == v)
        mminus = sum(1 for (u, v) in EX if g[u] == v and g[v] == u)
        Fe_p = Fe_m = 0
        if mplus:
            l = list(lens); l.remove(1); l.remove(1)
            Fe_p = mplus * coeff(P_from(l), K-1)
        if mminus:
            l = list(lens); l.remove(2)
            Fe_m = mminus * coeff(P_from(l), K-1)
        tot += Fe_p - Fe_m - Fv
    assert tot % len(G) == 0, "Burnside sum not divisible"
    return 1 + tot // len(G)

# ---- reader bridge ----
import itertools, numpy as np
from collections import defaultdict
ATOL=1e-9
def rank_abs(M):
    if M.size==0 or min(M.shape)==0: return 0
    return int((np.linalg.svd(M,compute_uv=False)>ATOL).sum())
def colspace(M):
    if M.size==0 or min(M.shape)==0: return np.zeros((M.shape[0],0))
    U,s,_=np.linalg.svd(M,full_matrices=False); return U[:,s>ATOL]

def build(adj,n,K,G):
    verts=sorted((frozenset(s) for s in itertools.combinations(range(n),K)),key=lambda s:sorted(s))
    vi={v:i for i,v in enumerate(verts)}
    emap={}   # undirected arena edge -> (index, canonical (u,w) meaning token moves u->w from A to B)
    edges=[]
    for x in verts:
        for u in x:
            for w in adj[u]:
                if w in x: continue
                y=frozenset((x-{u})|{w})
                key=frozenset((x,y))
                if key in emap: continue
                emap[key]=len(edges); edges.append((x,y,u,w))
    ei={}
    for i,(a,b,u,w) in enumerate(edges): ei[(a,b)]=(i,1); ei[(b,a)]=(i,-1)
    D0=np.zeros((len(edges),len(verts)))
    for (a,b,u,w) in edges:
        i,_=ei[(a,b)]; D0[i,vi[b]]+=1; D0[i,vi[a]]-=1
    P=np.zeros((len(edges),len(edges)))
    for g in G:
        for (a,b,u,w) in edges:
            i,_=ei[(a,b)]
            j,s=ei[(frozenset(g[e] for e in a),frozenset(g[e] for e in b))]
            P[j,i]+=s/len(G)
    # pullback of base 1-forms: nu_{u->w}(arena edge) = +1 if that edge moves u->w
    EX=edges_of(adj); W=np.zeros((len(edges),len(EX)))
    for i,(a,b,u,w) in enumerate(edges):
        k=EX.index(tuple(sorted((u,w))))
        W[i,k]= 1.0 if u<w else -1.0
    return verts,edges,D0,P,W,EX

def report(name,adj,n,K,G):
    verts,edges,D0,P,W,EX=build(adj,n,K,G)
    Inv=colspace(P); Ex=colspace(P@colspace(D0))
    R=Inv-Ex@(Ex.T@Inv); R=colspace(R)             # residual sector, dim = eps
    PW=colspace(P@colspace(W))                      # invariant part of pullback space
    reach_inv = rank_abs(R.T@PW) if PW.shape[1] else 0
    reach_raw = rank_abs(R.T@colspace(W))           # before averaging
    print(f"{name:<14} K={K}  V={len(verts):<4} E={len(edges):<5} "
          f"eps={R.shape[1]:<3} dim pi*C1(X)={rank_abs(W):<3} "
          f"inv part={PW.shape[1]:<3} | residual reached by readers: raw={reach_raw} averaged={reach_inv}")

# ---- aliases used by the reader-bridge helpers ----
cyc        = lambda q: C(q)
path       = lambda q: P(q)
hypercube  = lambda d: [(v, v ^ (1 << i)) for v in range(1 << d) for i in range(d) if v < v ^ (1 << i)]
petersen   = lambda: PET
prism      = lambda: PRISM
kbip       = lambda a, b: [(i, a + j) for i in range(a) for j in range(b)]
asym6      = lambda: [(0,1),(1,2),(2,3),(3,4),(4,5),(1,4),(2,5)]

def adjacency(E, n):
    a = defaultdict(set)
    for u, v in E: a[u].add(v); a[v].add(u)
    return {v: a[v] for v in range(n)}

PET_V = PV
PET_I = PI
```

### 9.2 $\epsilon$, its closed form, and the reader bound

### 9.3 The capacity filtration


```python
def ladder(nm, E, n, m, K):
    adj = adjacency(E, n)
    G = [g for g in autos(E, n) if g[m] == m]
    triv = [tuple(range(n))]
    print(f"{nm}  m={m}  K={K}  |Stab(m)|={len(G)}")
    for r in range(1, K):
        dRa, dRb, ck, ker = defect(adj, n, K, m, r, G)
        nv, ne, c, b1, lv = quotient(adj, n, K, m, r, G)
        _, _, _, ord_ker = defect(adj, n, K, m, r, triv)
        print(f"   r={r}->{r+1}: dim R {dRb}->{dRa}  coker={ck}  D_r={ker:<3} "
              f"| Q_r: V={nv} E={ne} c={c} beta1={b1:<3} | ordinary={ord_ker:<3} L={ord_ker-ker}"
              f"  {'ok' if ker==b1 else 'MISMATCH'}")

ladder("C7", C(7), 7, 0, 4)
ladder("Q3", hypercube(3), 8, 0, 4)
ladder("prism", PRISM, 6, 0, 4)
ladder("P6 (interior site)", P(6), 6, 2, 4)
```

### 9.4 Top layer, and the search for $c>1$


```python
print("top layer:  D_(K-1) = #(N(m)/Stab(m)) - 1")
tot = bad = 0
for nm, E, n in FAM:
    adj = adjacency(E, n); A = autos(E, n); seen = set()
    for m in range(n):
        if m in seen: continue
        seen |= {g[m] for g in A}
        G = [g for g in A if g[m] == m]
        orbs, sn = 0, set()
        for v in adj[m]:
            if v in sn: continue
            sn |= {g[v] for g in G}; orbs += 1
        for K in (3, 4):
            if K >= n+1 or not cap_arena(adj, n, K, m, K)[0]: continue
            _, _, _, ker = defect(adj, n, K, m, K-1, G)
            tot += 1; bad += (ker != orbs-1)
            if ker != orbs-1: print("   MISMATCH", nm, m, K, ker, orbs-1)
            elif orbs-1: print(f"   nonzero: {nm} m={m} K={K} deg={len(adj[m])} "
                               f"|Stab|={len(G)} orbits={orbs} -> D={ker}")
print(f"  {tot-bad}/{tot} agree")

print("\nsaturation K=|V|+r forces the old arena empty:")
for nm, E, n in FAM[:5]:
    adj = adjacency(E, n)
    for r in (1, 2):
        print(f"  {nm:<9} |V|={n} r={r} K={n+r}: |X^(r)| = {len(cap_arena(adj,n,n+r,0,r)[0])}")
```

## 10. Boundary of the result

These are not disclaimers. Each records a distinction established during the derivation, and each
prevents two mathematically different objects from being identified because they behaved similarly
on some slice of the parameter space. The boundary is part of the content.

$D_r \neq$ **reader information loss.** $D_r$ is a relative invariant cohomology group. Calling the
difference $L$ of Section 6 "loss" requires an independently defined observation map; Theorem 3.1
shows the readers actually constructed here see $\epsilon(X,1)$ of it and no more.

$D_r \neq$ **mass, energy, or price.** No dimensional map has been constructed. $p_r$ is an integer
sequence of Betti numbers.

$D_{K-1} = 0 \neq$ **price exhaustion.** Section 5.3 gives the exact cause: local stabiliser
transitivity at $m$.

**New states $\neq$ new channels.** For $C_7$ at $K=4$ the layers add $15, 6, 1$ configurations, carry
$25, 9, 1$ ordinary relative cycles, and contribute $11, 4, 0$ invariant channels. Three different
quantities; only the last is $p_r$.

**Symmetry filtering $\neq$ destruction.** $R_{G'} \hookrightarrow R_G$ is injective. What a larger
group cannot represent invariantly is still present.

**Change $\neq$ loss.** Loss $=$ change $+$ removal of admissible continuation, and even then the
removed sector is a relative class, not an annihilation.

$\epsilon$ **is not a base-graph invariant.** It is not controlled by degree, bipartiteness, $|V|$,
$|E|$, or girth. $C_6$ (bipartite) gives 0, $C_8$ (bipartite) gives 2, Petersen (non-bipartite)
gives 1. It measures how much loop structure survives the quotient — a property of the group's grip
on the transformation set.

**There is still no summary formula for $\epsilon(C_N,K)$.** The closed form of Theorem 2.3 computes
it; nothing compresses it.

## 11. Interpretation

*This section proves nothing and is separable from the rest of the paper.*

The results have a consistent shape. Every elementary change to the substrate or the observer either
adds transformation structure, re-expresses it, or removes it from view. The only operation that
produces a kernel is a restriction of which states are admissible, and the kernel it produces is a
relative class: the transformations are present in the larger admissible space and absent from the
smaller one. In this setting, therefore:

$$\text{change does not entail loss; loss} = \text{change} + \text{removal of admissible continuation.}$$

If one is looking for a notion of cost attached to transformation, the candidate this work produces
is not a scalar. It is the graded sequence $p_r = \beta_1(\mathcal Q_r)$ indexed by how much
admissibility has been relaxed, and its value is not the number of states admitted but the number
of independent transformation cycles they carry after exact and symmetry-trivial structure has been
quotiented away. Whether that has any physical referent is not addressed here.

The programme this belongs to holds that distinction requires lawful continuation and that no state
is terminal. The mathematics above is consistent with that and does not establish it. What it does
establish is narrower and, for that reason, usable: in a fully specified finite model of tokens
moving on a graph, the structure of transformation is conserved under every change of the rules of
movement except a restriction of what may exist, and that exception is measured exactly.

2. Find $\mathcal H^1_G \neq 0$, or prove it cannot occur. This is the single result that would
   reopen a non-coexact channel. Candidates: high-girth substrates with small automorphism groups.
3. A recurrence for $p_r$. Setting $k = K-r-1$, the layer's internal part is the ordinary $k$-token
   arena on $X - \{m\}$ with a cone attached wherever $N(m) \not\subseteq U$, so

   the natural next target.

   forbidden moves, or capacity below one elsewhere). Proposition 5.6 shows the present rules cannot
   produce one.
