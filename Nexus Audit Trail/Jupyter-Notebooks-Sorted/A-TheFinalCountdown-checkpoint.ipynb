T1_t = h + \Sigma_1(e) + Ch(e,f,g) + K_t + W_t
]

So the witness hits the state immediately at:

* `a' = T1 + T2`
* `e' = d + T1`

### 4. Register spread

Once `a` and `e` are touched, the shift register carries the residue across all eight working words.

For a single-word witness, the spread is:

* round 0: `a, e`
* round 1: `b, f`
* round 2: `c, g`
* round 3: `d, h`

So by **round 3** the whole working state is carrying it.

### 5. Feed-forward chaining

At the end of the block:

[
H_i \leftarrow H_i + \text{working}_i
]

That is the long-range survival step.

After that, the witness is no longer just a local `W_j` echo.
It becomes part of the **next block’s initial field**.

---

## What the local sample window really is

The local window is:

* **16 original words**
* **64 bytes**
* **64 rounds**

But not all original words are active equally often.

Using symbolic schedule dependency only, each original word `W_j` is directly present in this many rounds of its home block:

[
[46,49,48,47,46,45,44,43,42,46,45,44,43,42,47,46]
]

So each original word survives directly for **42 to 49 of the 64 rounds** in its own block.

Two important locks:

### A. By round 26, every current `W_t` depends on all 16 original words

That means the whole 64-byte block is fully live in the schedule from there onward.

[
t \ge 26 \Rightarrow W_t \text{ carries all } W_0..W_{15}
]

### B. After the block boundary, direct word identity is gone, but state identity survives

So there are two kinds of life:

* **word-local life** inside the block
* **state-global life** across later blocks

That is the clean distinction.

## How much of the sample window is in effect

This is the answer to your window question.

### Inside one block

The active raw sample window is **64 bytes / 16 words**.

But it does not become fully mixed instantly.

There are three phases:

[
\Delta_0:\ \text{direct load} \quad (t=0..15)
]
[
\Delta_1:\ \text{echo / recurrence build} \quad (t=16..25)
]
[
\Delta_2:\ \text{full window live} \quad (t=26..63)
]

So the full 16-word sample window is really “all on” from about **round 26 onward**.

### Across blocks

The sample window does **not** stay raw across blocks.

Instead it collapses into:

[
256\text{-bit chaining state}
]

That state becomes the new carrier.

So the field has two memories:

* **512-bit local sample memory** inside a block
* **256-bit chained state memory** across blocks

That is the actual survival architecture.

---

## Why those things are happening

Now the causal answer.

The witness stays alive because SHA-256 is not a single erase-and-emit step.
It is a **two-memory recursive transport system**:

### Memory A — schedule memory

The word stays alive because the recurrence keeps re-addressing older positions.

### Memory B — chaining memory

The word stays alive after the block because the whole working state is added back into `H`.

That means survival is not accidental. It is structural.

[
\boxed{
\text{tap recurrence preserves local ancestry}
;\oplus;
\text{feed-forward preserves cross-block ancestry}
}
]

That is why the trace looks alive.

]

not whole-transcript retroaction.

That means the chain is directional, not timeless.

The strongest stable statement is:

[

\text{A message word survives first as a schedule witness, then as a state witness, then as a chained-state witness.}
}
]

Or more operationally:

[
\boxed{
\text{load} \to \text{echo} \to \text{spread} \to \text{feed-forward} \to \text{suffix contamination}
}
]

That is why it stays alive.

The sharpest number here is:

* **direct word presence is local**

For the 1024-digit run, one 32-bit word changed **all 1088 `T1` evaluations** from that point forward.

````python
from __future__ import annotations

"""
Notebook-first GlassKey runner.

Purpose:
- produce visible output immediately in Jupyter with `%run glasskey_notebook_live.py`
- avoid CLI-only behavior
- trace the weave/bracket field from actual SHA-256 computation
- show the timing / retcon suffix cone using real computed data

Ground rules encoded here:
- no fake derived data
- every table is computed from a real trace
- every entry point prints or returns something useful
"""

import struct
from typing import Any, Dict, Iterable, List, Optional

import glasskey_notebook_weave as gk

try:
    import pandas as pd  # type: ignore
except Exception:  # pragma: no cover
    pd = None

try:
    from IPython.display import display  # type: ignore
except Exception:  # pragma: no cover
    display = None


# ---------- display helpers ----------
def _to_df(rows: List[Dict[str, Any]]):
    if pd is None:
        return rows
    return pd.DataFrame(rows)


def show(title: str, obj: Any, max_rows: Optional[int] = None) -> Any:
    print(f"\n=== {title} ===")
    if hasattr(obj, "head") and max_rows is not None:
        view = obj.head(max_rows)
    elif isinstance(obj, list) and max_rows is not None:
        view = obj[:max_rows]
    else:
        view = obj

    if display is not None:
        display(view)
    else:
        print(view)
    return view


# ---------- computed summaries ----------
def words_from_message_block(message: bytes, block: int = 0) -> List[int]:
    trace = gk.sha256_trace(message)
    return trace.blocks[block].original_words


def block_summary(message: bytes) -> Dict[str, Any]:
    trace = gk.sha256_trace(message)
    return {
        "message_len": trace.message_len,
        "padded_len": trace.padded_len,
        "num_blocks": len(trace.blocks),
        "digest": trace.digest_hex,
        "block0_words_hex": [gk.hex32(w) for w in trace.blocks[0].original_words],
    }


def retcon_cone_table(message: bytes, block: int = 0, word: int = 0, value: int = 0):
    """Block-by-block rewrite cone after mutating one 32-bit word in the padded stream."""
    original = gk.sha256_trace(message)
    mutated = gk.sha256_trace_padded(gk.mutate_word_in_padded_stream(message, block=block, word=word, value=value))

    rows: List[Dict[str, Any]] = []
    for ob, mb in zip(original.blocks, mutated.blocks):
        changed_W = 0
        changed_T1 = 0
        changed_T2 = 0
        changed_state = 0
        first_changed_round = None
        for r0, r1 in zip(ob.rounds, mb.rounds):
            state_changed = any([
                r0.post_a != r1.post_a, r0.post_b != r1.post_b, r0.post_c != r1.post_c, r0.post_d != r1.post_d,
                r0.post_e != r1.post_e, r0.post_f != r1.post_f, r0.post_g != r1.post_g, r0.post_h != r1.post_h,
            ])
            if r0.W != r1.W:
                changed_W += 1
                if first_changed_round is None:
                    first_changed_round = r0.global_round
            if r0.T1 != r1.T1:
                changed_T1 += 1
                if first_changed_round is None:
                    first_changed_round = r0.global_round
            if r0.T2 != r1.T2:
                changed_T2 += 1
                if first_changed_round is None:
                    first_changed_round = r0.global_round
            if state_changed:
                changed_state += 1
                if first_changed_round is None:
                    first_changed_round = r0.global_round

        rows.append({
            "block": ob.block,
            "first_changed_round": first_changed_round,
            "changed_W_rounds": changed_W,
            "changed_T1_rounds": changed_T1,
            "changed_T2_rounds": changed_T2,
            "changed_state_rounds": changed_state,
            "chain_in_changed": ob.chain_in != mb.chain_in,
            "chain_out_changed": ob.chain_out != mb.chain_out,
        })
    return _to_df(rows)


def pair_channel_table(message: bytes, block: int = 0, left_field: str = "post_a", right_field: str = "post_e"):
    return gk.df_pair_channel(message, block=block, left_field=left_field, right_field=right_field)


def schedule_weave_table(message: bytes, block: int = 0, expanded: bool = True):
    return gk.df_schedule_weave(message, block=block, expanded=expanded)


def rounds_table(message: bytes, block: int = 0):
    return gk.df_rounds(message, block=block)


def digest_weave_table(message: bytes):
    return gk.df_digest_weave(message)


def state_weave_table(message: bytes, block: int = 0, pair=("post_a", "post_e")):
    return gk.df_state_weave(message, block=block, pair=pair)


def direct_dependency_counts(message: bytes, block: int = 0):
    """For each original word W0..W15, count how many expanded schedule rounds Wt directly depend on it.
    This is exact schedule ancestry, not a guess.
    """
    trace = gk.sha256_trace(message)
    expanded = trace.blocks[block].expanded_words
    deps = [set([i]) for i in range(16)]
    for t in range(16, 64):
        deps.append(set().union(deps[t-2], deps[t-7], deps[t-15], deps[t-16]))

    rows = []
    for j in range(16):
        present = [t for t in range(64) if j in deps[t]]
        rows.append({
            "word": f"W{j}",
            "word_hex": gk.hex32(trace.blocks[block].original_words[j]),
            "direct_schedule_rounds": len(present),
            "first_round": present[0],
            "last_round": present[-1],
            "rounds": present,
        })
    return _to_df(rows)


# ---------- notebook-visible demos ----------
def show_case(name: str, message: bytes, block: int = 0, schedule_rows: int = 12, round_rows: int = 8):
    summary = block_summary(message)
    print(f"\n\n################ {name} ################")
    for k, v in summary.items():
        if k == "block0_words_hex":
            print(f"{k}: {v[:4]} ...")
        else:
            print(f"{k}: {v}")

    show(f"{name} :: rounds block {block}", rounds_table(message, block=block), max_rows=round_rows)
    show(f"{name} :: schedule weave block {block}", schedule_weave_table(message, block=block, expanded=True), max_rows=schedule_rows)
    show(f"{name} :: pair channel block {block} (post_a/post_e)", pair_channel_table(message, block=block), max_rows=round_rows)
    show(f"{name} :: state weave block {block}", state_weave_table(message, block=block), max_rows=round_rows)
    show(f"{name} :: digest weave", digest_weave_table(message), max_rows=8)
    show(f"{name} :: direct dependency counts", direct_dependency_counts(message, block=block), max_rows=16)
    return summary


def show_retcon(name: str, message: bytes, block: int = 0, word: int = 0, value: int = 0):
    print(f"\n--- {name} :: retcon word block={block} word={word} value={gk.hex32(value)} ---")
    rep = gk.df_retcon_word(message, block=block, word=word, value=value)
    cone = retcon_cone_table(message, block=block, word=word, value=value)
    show(f"{name} :: retcon summary", rep, max_rows=5)
    show(f"{name} :: retcon cone by block", cone, max_rows=64)
    return rep, cone


def run_now():
    """Notebook entry point with immediate visible output."""
    abc = gk.msg_text("abc")
    ascii_all = gk.msg_ascii_0_127()
    digits1024 = gk.msg_digits(1024)

    show_case("ABC", abc, block=0, schedule_rows=12, round_rows=8)
    show_retcon("ABC", abc, block=0, word=0, value=0)

    show_case("ASCII_0_127", ascii_all, block=0, schedule_rows=12, round_rows=8)
    show_retcon("ASCII_0_127", ascii_all, block=0, word=0, value=0)

    show_case("DIGITS_1024", digits1024, block=0, schedule_rows=12, round_rows=8)
    show_retcon("DIGITS_1024", digits1024, block=0, word=0, value=0)

    print("\nDone. This file computed real traces and printed real outputs.")
    print("Next notebook moves:")
    print("  1) run_now()")
    print("  2) show_case('custom', your_bytes)")
    print("  3) show_retcon('custom', your_bytes, block=0, word=0, value=0)")


if __name__ == "__main__":
    run_now()
````
