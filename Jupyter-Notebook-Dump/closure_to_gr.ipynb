## 3. Newtonian Limit — Inverse-Square Recovery

```python
# ── Weak-field, static, slow-motion limit ───────────────────────────────────
#
# Metric perturbation:  g_{00} ≈ -(1 + 2Φ/c²),  g_{ij} ≈ (1 - 2Φ/c²)δ_{ij}
#
# Einstein eq → G_{00} = κ T_{00}  →  Poisson:
#   ∇²Φ = 4πG ρ_Ψ
#
# Outside a spherical source (ρ_Ψ = 0):
#   ∇²Φ = 0,  spherical symmetry  →  Φ(r) = -GM/r

Phi_r = -G * M / r
accel = -diff(Phi_r, r)

print('Gravitational potential (vacuum, spherical):')
display(Math(r'\Phi(r) = ' + latex(Phi_r)))
print()
print('Acceleration a = -∇Φ:')
display(Math(r'a = ' + latex(accel) + r'\;\hat{r}'))
print()
print('→ Inverse-square law recovered from spherical vacuum geometry.')

# ── Verify: ∇²Φ = 0 in spherical coords outside source ──────────────────────
# Laplacian in 3D spherical: (1/r²) d/dr(r² dΦ/dr)
laplacian_Phi = (1/r**2) * diff(r**2 * diff(Phi_r, r), r)
laplacian_Phi_simplified = simplify(laplacian_Phi)
print(f'∇²Φ = {laplacian_Phi_simplified}  (vanishes outside source, as required)')
```

## 5. Gravitational Redshift

```python
# ── Static metric, proper time and frequency ratio ───────────────────────────
#
# dτ = √(-g_{00}) dt
# ν_obs / ν_emit = √(g_{00}(emit) / g_{00}(obs))
#
# Weak-field: g_{00} ≈ -(1 + 2Φ/c²)

Phi_emit, Phi_obs = symbols('Phi_emit Phi_obs', real=True)

g00_emit = -(1 + 2*Phi_emit/c**2)
g00_obs  = -(1 + 2*Phi_obs /c**2)

# Exact ratio
freq_ratio_exact = sqrt(g00_emit / g00_obs)

# Weak-field expansion: Φ/c² ≪ 1
# (1 + 2Φ_emit/c²) / (1 + 2Φ_obs/c²) ≈ 1 + 2(Φ_emit - Φ_obs)/c²
delta_Phi = symbols('DeltaPhi', real=True)
ratio_approx = 1 + delta_Phi / c**2   # from series expansion

print('Exact frequency ratio:')
display(Math(
    r'\frac{\nu_{\rm obs}}{\nu_{\rm emit}}'
    r'= \sqrt{\frac{g_{00}(\rm emit)}{g_{00}(\rm obs)}}'
))
print()
print('Weak-field limit (ΔΦ = Φ_emit - Φ_obs):')
display(Math(
    r'\frac{\Delta\nu}{\nu} \approx \frac{\Delta\Phi}{c^2}'
))
print()
print('Redshift recovered because stored closure alters clocking through g_{00}.')
```
