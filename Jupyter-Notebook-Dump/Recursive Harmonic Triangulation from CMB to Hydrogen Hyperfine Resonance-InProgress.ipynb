## 3. Planckian Peak Frequency

### 3.1 Derivation of $\nu_{\mathrm{peak}}$

The Planck spectral radiance per unit frequency is

$$
B_\nu(T)=\frac{2h\nu^3}{c^2}\frac{1}{e^{h\nu/(k_BT)}-1}\,.
$$

Setting $dB_\nu/d\nu=0$ yields the transcendental equation

$$
3(1-e^{-x})=x,\quad x=\frac{h\nu_{\mathrm{peak}}}{k_BT}\,,\quad x\approx2.821439.
$$

Consequently,

$$
\nu_{\mathrm{peak}}=\frac{x\,k_BT_{\mathrm{CMB}}}{h}
=2.821439\frac{k_BT_{\mathrm{CMB}}}{h}\approx1.60\times10^{11}\,\mathrm{Hz}
\;(160\,\mathrm{GHz}).
$$

$$
