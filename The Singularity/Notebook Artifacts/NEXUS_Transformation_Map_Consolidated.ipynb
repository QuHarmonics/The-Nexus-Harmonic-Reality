## Portable map of the architecture

The current model separates six different objects rather than collapsing them into one:

1. **Route word (NXRT):** the ordered, typed transformation instructions. It is fixed independently of the runtime payload.
2. **Composed operator (NXOP):** the canonical action produced by composing the route. Multiple different histories can occupy the same operator groove.
3. **Realized trace (NXTRACE):** payload-conditioned branches/parameters actually selected during a run. It is not the static route program.
4. **Surface/readout:** the value or state exposed after a projection. Distinct inputs/routes can collide at this layer.
5. **Observer/residual:** fixed inverse rules plus any necessary within-fiber address, recursively composed into one root path back. A residual is not the observer program itself.
6. **Context wrapper:** the higher-level relation that places a transformation in a broader domain. It is not silently inserted into NXRT or NXOP.

The compact witness is

\[
\boxed{5=2+3=5}
\]

Read as three typed positions: target surface, admissible route, verified landing. The first equality pulls a route compatible with the target; the second closes when the route lands on that same target. Erase the middle and the route/history disappears into `5 = 5`.

## Core laws kept separate

- **Route identity is not operator identity.** Distinct route histories can compose to the same operator (“collision of logic” / groove).
- **Operator identity is not output identity.** Different payloads can collide under a many-to-one readout (“collision of matter” / leaving the groove).
- **Reversibility belongs to transformation logic plus its return information, not to a folded scalar value.** If a transform merges states, the lost distinction must be retained in a residual/held lane or it cannot be recovered from the value alone.
- **Prime addresses encode position; grammar defines meaning.** \(G(r)=\prod_i p_i^{t_i+1}\) is injective for a well-framed nonnegative token sequence, including token ID zero and trailing zero-valued tokens. Unique factorization guarantees decoding, but does not invent the instruction semantics.
- **SHA-256 is a boundary case, not a shortcut around the observer problem.** A round is reversible when its schedule word and constant are known; the complete digest is not an inverse interface. Feed-forward occurs at every compression block. The observer-sidecar tests return inputs because they retain message-derived route information; that is not inversion from the digest alone.
- **Gradient/path rule.** With an anchor \(x_0\) and ordered increments \(g_i=x_{i+1}-x_i\), the route is reconstructed by prefix sums. The endpoint \(x_n-x_0=\sum_i g_i\) is only a projection of that route.

### 99.1 What a handle destroys is exactly order — `noloss.py`

Two different maps get called addition. Keep the carry on its own wire and the deficit is exact:

```
w= 1  carry kept (x^y, x&y): outputs      3 = 3^w   bits lost 0.500  = w/2
w= 2                                      9         bits lost 1.000  = w/2
w= 3                                     27         bits lost 1.500  = w/2
w= 4                                     81         bits lost 2.000  = w/2
w= 8                                   6561         bits lost 4.000  = w/2
w= 8  carry folded (x+y):   outputs    511         bits lost 7.279
```

Every fiber is exactly one swap orbit — 81 of 81 checked at w = 4, zero exceptions, fiber size always
2^(number of differing bits). So the only thing the handle can destroy is **which side held the 1**.
Adding one bit per differing position restores the bijection exactly at every width tested (4, 16, 64,
256, 65 536 images of the same counts). Field-side the handle loses nothing; what it loses is the
traversal. RUNS.

Folding the carry into one channel costs more and loses structure. At w = 3, target 5:

```
(xor=1, and=2) <- (2,3) (3,2)
(xor=5, and=0) <- (0,5) (1,4) (4,1) (5,0)
```

Two routes kept against six folded. The pair separates 2+3 from 1+4 from 0+5; the number 5 does not.
So *you always know the path back* holds exactly while the carry has its own wire. The loss is not in
the arithmetic; it is in reusing one channel for two things. `5 = 2 + 3` is lossless because the
equals sign **is** the kept carry: an equation holds the route, an evaluation folds it. This is § 77's
location result read from the cost side (recovered, Nerode).

**Order is a group, not a quantity** — correction to my own first unification of 99.1 and 99.2. At
w = 6 the half adder forgets (Z/2)^w pointwise and loses 3.0000 bits (= w/2); a route forgets S_n and
loses 3.6666. Different forgetting groups, both reader-side, deficit = log of the orbit. "Order" carries
a type.

### 99.6 Rings: the stack is a maximal chain of cuts, and no complete chain is order-independent — `rings.py`, `rings2.py`

The witness field, order 24, 26 subgroups of orders 1, 2, 3, 4, 6, 8, 12, 24.

```

chain lengths (= number of measurements): {3: 12, 4: 24}
every chain's indices multiply to the field size: True
example long  [2, 3, 2, 2]   example short [4, 3, 2]

proper subgroup pairs whose two cuts do not commute: 108 of 276
```

So the diagnostic stack is a chain of covers with kept × leftover = whole at every step, and its depth is
**not fixed by the field**: 3 or 4 depending on which cut is taken. That is why a selection rule is needed
rather than hardcoded, and the lattice calculates it — the indices of the six covers are visible before
any measurement, and taking the largest first gives [4,3,2] instead of [2,3,2,2]. Accumulation and fitting
are one lattice walked in opposite directions, and the noise is structural, not sloppiness: going down,
every step is forced to be a cover with its index known in advance; going up, one added element reaches 16
different subgroups and the jump is known only afterwards.

```
length | chains | every step order-independent | passes through a NON-normal cut

```

**Zero chains are order-independent, at either length.** The order-2 normal subgroup is the central factor

### 99.7 Why k = 2, routed rather than searched

C1a says independent distinctions must remain independently continuable, [T_A, T_B] = 0 — **independent
means commuting**. Commuting generators generate an abelian monoid, which is already its own abelianization:
fiber size 1, no order to forget. So the surface is not a projection that happens to lose order; the surface
is **the act of declaring the generators independent**, and the fiber is what that declaration costs. k counts
the independent directions and nothing else.

The base admits two readings — the whole, complete and at rest; the part, partial and therefore in motion —
and which one is got depends only on whether the reader is the whole or a piece of it, with no piece ever the
whole. A reading is the only thing that can serve as an independent direction, so k = 2. BOUND, one provider.

Held together, both spaces at once: the surface pretends the two readings are two things, the inside knows they
are one base read twice, and **the order channel is the residue of that pretence**. Run backwards — if the
generators genuinely commuted at the inside level the fiber would be 1, the surface would equal the inside, and
there would be no order, no undetermined continuation, no motion: the whole, at rest. Fiber size measures how
far a reader is from being the whole.

**Flagged, not claimed.** "Two is the minimum and there is exactly one of it" is inherited from the entry, not
re-run here. The exhaustiveness of the two readings is the domain asserting it — a base fact, not a theorem
executed. And the atomic application cannot be used to pick k = 2 and then be presented as proof that k = 2 was
forced.

**Correction to §107, logged.** Exact irrational root independence does not make the truncated 32-bit words linearly independent, nor does it by itself establish cryptographic independence.
````python
def add_hold(a: int, b: int) -> tuple[int, int]:
    """Carry-retaining addition: update sum, preserve b as return coordinate."""
    return a + b, b

def add_hold_inverse(s: int, b: int) -> tuple[int, int]:
    return s - b, b

# Two different matter configurations, same scalar readout.
assert 3 + 2 == 4 + 1 == 5
assert add_hold(3, 2) == (5, 2)
assert add_hold(4, 1) == (5, 1)
assert add_hold_inverse(*add_hold(3, 2)) == (3, 2)
assert add_hold_inverse(*add_hold(4, 1)) == (4, 1)

# Different histories can land in the same operator groove: identity vs swap twice.
def swap_pair(x):
    a, b = x
    return b, a
states = [(0,0), (0,1), (1,0), (1,1)]
assert all(swap_pair(swap_pair(s)) == s for s in states)
assert all(s == s for s in states)  # empty route is identity too

# Frame condition: T(a,b,c)=(a+b,b,c) does not rewrite the unowned c lane.
def T_frame(a, b, c):
    return a + b, b, c
frame_checks = 0
for a in range(-2, 3):
    for b in range(-2, 3):
        for c in range(-2, 3):
            assert T_frame(a,b,c)[2] == c
            assert T_frame(a,b,c)[1] == b
            frame_checks += 1
print({"same_scalar_sum": 5, "different_route_preserving_states": [(5,2),(5,1)],
       "inverse_returns": True, "frame_states_checked": frame_checks})
````

## Test 2 — Prime-address route encoding preserves zero tokens and positions

````python
from itertools import product
from math import isqrt

def primes_up_to_count(n):
    out, candidate = [], 2
    while len(out) < n:
        if all(candidate % p for p in out if p * p <= candidate):
            out.append(candidate)
        candidate += 1
    return out

def encode_route(tokens):
    ps = primes_up_to_count(len(tokens))
    code = 1
    for p, token in zip(ps, tokens):
        if not isinstance(token, int) or token < 0:
            raise ValueError("route tokens must be nonnegative integers")
        code *= p ** (token + 1)
    return code

def factor_exponents(n):
    if n < 1:
        raise ValueError("code must be positive")
    factors = {}
    d = 2
    while d*d <= n:
        while n % d == 0:
            factors[d] = factors.get(d, 0) + 1
            n //= d
        d = 3 if d == 2 else d + 2
    if n > 1:
        factors[n] = factors.get(n, 0) + 1
    return factors

def decode_route(code):
    if code == 1:
        return ()
    factors = factor_exponents(code)
    pos_primes = primes_up_to_count(len(factors))
    if set(factors) != set(pos_primes):
        raise ValueError("not a canonical consecutive-position route code")
    return tuple(factors[p] - 1 for p in pos_primes)

routes6 = [r for n in range(6) for r in product(range(6), repeat=n)]
codes6 = [encode_route(r) for r in routes6]
assert len(routes6) == 9331 == len(set(codes6))
assert all(decode_route(c) == r for r, c in zip(routes6, codes6))
assert encode_route((4, 0, 5, 0)) != encode_route((4, 0, 5))  # trailing zero-valued token retained

# The unshifted exponent idea fails when token 0 has exponent zero.
def encode_unshifted(tokens):
    ps = primes_up_to_count(len(tokens))
    code = 1
    for p, token in zip(ps, tokens):
        code *= p ** token
    return code
unshifted_collision = encode_unshifted((1,)) == encode_unshifted((1,0))
assert unshifted_collision
print({"routes_length_0_to_5_over_6_tokens": len(routes6),
       "unique_codes": len(set(codes6),), "all_decoded": True,
       "trailing_zero_position_preserved": True, "unshifted_collision_witness": unshifted_collision})
````

## Test 3 — Recursive observer as a within-fiber address

````python
from math import ceil, log2

# Surface fold H(a,b)=a XOR b on two 4-bit lanes.
fibers = {h: [] for h in range(16)}
returns = 0
for a in range(16):
    for b in range(16):
        h = a ^ b
        residual = a  # address of the member within the fiber
        fibers[h].append((a,b))
        recovered = (residual, h ^ residual)
        assert recovered == (a,b)
        returns += 1
sizes = [len(fibers[h]) for h in range(16)]
residual_bits = ceil(log2(max(sizes)))
assert returns == 256 and set(sizes) == {16} and residual_bits == 4
print({"source_pairs": returns, "surface_values": len(fibers),
       "fiber_sizes": sorted(set(sizes)), "minimum_fixed_width_residual_bits": residual_bits,
       "exact_returns": returns})
````

## Test 4 — A gradient is an ordered step in a larger route

````python
def increments(path):
    return [path[i+1] - path[i] for i in range(len(path)-1)]

def integrate_steps(anchor, steps):
    out = [anchor]
    for step in steps:
        out.append(out[-1] + step)
    return out

path_a = [10, 12, 15, 14, 20]
steps_a = increments(path_a)
assert integrate_steps(path_a[0], steps_a) == path_a

# Same endpoints, different ordered route: endpoint delta alone folds the histories.
path_b = [10, 13, 17, 20]
steps_b = increments(path_b)
assert path_a[0] == path_b[0] and path_a[-1] == path_b[-1]
assert sum(steps_a) == sum(steps_b) == 10
assert steps_a != steps_b
print({"path_A": path_a, "steps_A": steps_a, "path_B": path_b, "steps_B": steps_b,
       "same_endpoint_delta": sum(steps_a), "same_route": steps_a == steps_b,
       "reconstruction_A_exact": integrate_steps(10, steps_a) == path_a})
````

## Test 5 — SHA-256 round inversion when schedule inputs are retained

````python
import hashlib, random, struct
MASK32 = 0xffffffff
SHA_IV = (0x6a09e667,0xbb67ae85,0x3c6ef372,0xa54ff53a,0x510e527f,0x9b05688c,0x1f83d9ab,0x5be0cd19)
SHA_K = (
0x428a2f98,0x71374491,0xb5c0fbcf,0xe9b5dba5,0x3956c25b,0x59f111f1,0x923f82a4,0xab1c5ed5,
0xd807aa98,0x12835b01,0x243185be,0x550c7dc3,0x72be5d74,0x80deb1fe,0x9bdc06a7,0xc19bf174,
0xe49b69c1,0xefbe4786,0x0fc19dc6,0x240ca1cc,0x2de92c6f,0x4a7484aa,0x5cb0a9dc,0x76f988da,
0x983e5152,0xa831c66d,0xb00327c8,0xbf597fc7,0xc6e00bf3,0xd5a79147,0x06ca6351,0x14292967,
0x27b70a85,0x2e1b2138,0x4d2c6dfc,0x53380d13,0x650a7354,0x766a0abb,0x81c2c92e,0x92722c85,
0xa2bfe8a1,0xa81a664b,0xc24b8b70,0xc76c51a3,0xd192e819,0xd6990624,0xf40e3585,0x106aa070,
0x19a4c116,0x1e376c08,0x2748774c,0x34b0bcb5,0x391c0cb3,0x4ed8aa4a,0x5b9cca4f,0x682e6ff3,
0x748f82ee,0x78a5636f,0x84c87814,0x8cc70208,0x90befffa,0xa4506ceb,0xbef9a3f7,0xc67178f2)

def rotr32(x,n): return ((x >> n) | (x << (32-n))) & MASK32
def BSIG0(x): return rotr32(x,2) ^ rotr32(x,13) ^ rotr32(x,22)
def BSIG1(x): return rotr32(x,6) ^ rotr32(x,11) ^ rotr32(x,25)
def CH(x,y,z): return ((x & y) ^ ((~x) & z)) & MASK32
def MAJ(x,y,z): return (x & y) ^ (x & z) ^ (y & z)

def sha_round(state, w, k):
    a,b,c,d,e,f,g,h = state
    t1 = (h + BSIG1(e) + CH(e,f,g) + k + w) & MASK32
    t2 = (BSIG0(a) + MAJ(a,b,c)) & MASK32
    return ((t1+t2)&MASK32, a, b, c, (d+t1)&MASK32, e, f, g)

def sha_round_inverse(state2, w, k):
    ap,bp,cp,dp,ep,fp,gp,hp = state2
    a,b,c,e,f,g = bp,cp,dp,fp,gp,hp
    t2 = (BSIG0(a) + MAJ(a,b,c)) & MASK32
    t1 = (ap - t2) & MASK32
    d = (ep - t1) & MASK32
    h = (t1 - BSIG1(e) - CH(e,f,g) - k - w) & MASK32
    return (a,b,c,d,e,f,g,h)

# 2,000 exact inverse checks with W and K held fixed.
rng_round = random.Random(107)
round_checks = 2000
for _ in range(round_checks):
    state = tuple(rng_round.getrandbits(32) for _ in range(8))
    w, k = rng_round.getrandbits(32), rng_round.getrandbits(32)
    after = sha_round(state, w, k)
    assert sha_round_inverse(after, w, k) == state

# Small but useful independent digest oracle; this is not digest inversion.
sample_digest = hashlib.sha256(b'3+2').hexdigest()
assert len(sample_digest) == 64
print({"sample_message_sha256": sample_digest, "conditional_sha_round_inverse_checks": round_checks,
       "all_pass": True, "scope": "one SHA-256 round, conditioned on known W and K; not digest-only inversion"})
````

## Test 6 — Recursive SHA observer and exact scope boundary

````python
def sha_pad(msg: bytes) -> bytes:
    bit_len = len(msg) * 8
    out = msg + b'\x80'
    out += b'\x00' * ((56 - len(out) % 64) % 64)
    out += struct.pack('>Q', bit_len)
    return out

def small_sigma0(x): return rotr32(x,7) ^ rotr32(x,18) ^ (x >> 3)
def small_sigma1(x): return rotr32(x,17) ^ rotr32(x,19) ^ (x >> 10)

def expand_block(block):
    if len(block) != 64: raise ValueError('SHA block must be 64 bytes')
    w = list(struct.unpack('>16I', block))
    for t in range(16,64):
        w.append((small_sigma1(w[t-2]) + w[t-7] + small_sigma0(w[t-15]) + w[t-16]) & MASK32)
    return tuple(w)

def sha_rounds(state, words):
    for k,w in zip(SHA_K, words):
        state = sha_round(state, w, k)
    return state

def sha_inverse_rounds(state, words):
    for k,w in reversed(tuple(zip(SHA_K, words))):
        state = sha_round_inverse(state, w, k)
    return state

def words_to_bytes(words): return struct.pack('>8I', *words)
def bytes_to_words(data): return struct.unpack('>8I', data)

def forward_observer(msg):
    padded = sha_pad(msg)
    h = SHA_IV
    nodes = []
    for off in range(0, len(padded), 64):
        block = padded[off:off+64]
        words = expand_block(block)
        pre = h
        e = sha_rounds(pre, words)
        h = tuple((x+y) & MASK32 for x,y in zip(pre,e))
        # This is an explicit sidecar: it retains message-derived block and E.
        nodes.append({'block': block, 'E': e})
    return words_to_bytes(h), {'nodes': nodes}

def strip_sha_pad(padded):
    if not padded or len(padded) % 64: raise ValueError('invalid padded input')
    bit_len = struct.unpack('>Q', padded[-8:])[0]
    if bit_len % 8: raise ValueError('byte-aligned messages only')
    n = bit_len // 8
    if n + 1 > len(padded)-8 or padded[n] != 0x80: raise ValueError('padding marker mismatch')
    if any(padded[n+1:-8]): raise ValueError('padding fill mismatch')
    if len(padded) != ((n+9+63)//64)*64: raise ValueError('padding length mismatch')
    return padded[:n]

def observe_back(digest, observer):
    current = bytes_to_words(digest)
    recovered_blocks, inverse_count = [], 0
    for node in reversed(observer['nodes']):
        block, E = node['block'], node['E']
        words = expand_block(block)
        pre = tuple((x-y) & MASK32 for x,y in zip(current,E))
        assert sha_inverse_rounds(E, words) == pre
        inverse_count += 64
        recovered_blocks.append(block)
        current = pre
    assert current == SHA_IV
    msg = strip_sha_pad(b''.join(reversed(recovered_blocks)))
    return msg, inverse_count

rng_messages = random.Random(111)
fixed_messages = [b'', b'3+2', b'a'*55, b'a'*56, b'a'*63, b'a'*64, b'a'*65,
                  bytes(range(256)), b'z'*119, b'z'*120, b'z'*127, b'z'*128, b'z'*129,
                  bytes(rng_messages.randrange(256) for _ in range(1024))]
random_messages = [bytes(rng_messages.randrange(256) for _ in range(n))
                   for n in [2,17,54,57,62,127,128,255,512,777] for _ in range(4)]
messages = fixed_messages + random_messages
assert len(messages) == 54
hash_matches = exact_returns = total_blocks = total_inverse_rounds = 0
for msg in messages:
    digest, obs = forward_observer(msg)
    hash_matches += (digest == hashlib.sha256(msg).digest())
    recovered, inverse_calls = observe_back(digest, obs)
    exact_returns += (recovered == msg)
    total_blocks += len(obs['nodes'])
    total_inverse_rounds += inverse_calls
assert (hash_matches, exact_returns, total_blocks, total_inverse_rounds) == (54,54,207,13248)
print({"messages": len(messages), "hashlib_digest_matches": hash_matches,
       "observer_exact_returns": exact_returns, "compression_blocks": total_blocks,
       "inverse_round_applications": total_inverse_rounds,
       "scope": "observer sidecar retains each padded block and E; not inversion from digest alone"})
````

## Test 7 — Removing the redundant E trace is not input compression

````python
def schedule_from_tail(tail):
    if len(tail) != 16: raise ValueError('tail must hold W[48..63]')
    w = [None] * 64
    w[48:64] = list(tail)
    for t in range(63, 15, -1):
        assert all(w[j] is not None for j in (t, t-2, t-7, t-15))
        w[t-16] = (w[t] - small_sigma1(w[t-2]) - w[t-7] - small_sigma0(w[t-15])) & MASK32
    assert all(v is not None for v in w)
    return tuple(w)

def block_from_words(words): return struct.pack('>16I', *words[:16])

blocks_seen = schedule_roundtrips = stored_data_tails = padding_only_blocks = 0
input_bytes = 0
for msg in messages:
    input_bytes += len(msg)
    padded = sha_pad(msg)
    blocks = [padded[i:i+64] for i in range(0,len(padded),64)]
    data_block_count = (len(msg)+63)//64
    for block in blocks:
        full = expand_block(block)
        reconstructed = schedule_from_tail(full[48:64])
        assert reconstructed == full and block_from_words(reconstructed) == block
        schedule_roundtrips += 1
    blocks_seen += len(blocks)
    stored_data_tails += data_block_count
    padding_only_blocks += len(blocks) - data_block_count

legacy_bits = blocks_seen * 768                # 512-bit block + 256-bit E
reduced_tails_bits = stored_data_tails * 512    # W[48..63] recodes full block
length_metadata_bits = len(messages) * 64
input_plus_metadata_bits = input_bytes * 8 + length_metadata_bits
reduced_bits = reduced_tails_bits + length_metadata_bits
assert (blocks_seen, schedule_roundtrips, stored_data_tails, padding_only_blocks) == (207,207,174,33)
assert legacy_bits == 158976 and reduced_tails_bits == 89088 and reduced_bits == 92544
assert input_plus_metadata_bits == 84840
print({"messages":len(messages), "blocks":blocks_seen, "schedule_roundtrips":schedule_roundtrips,
       "data_bearing_tail_records":stored_data_tails, "padding_only_blocks_regenerated":padding_only_blocks,
       "previous_trace_bits":legacy_bits, "reduced_schedule_tail_plus_length_bits":reduced_bits,
       "input_bytes_plus_same_length_metadata_bits":input_plus_metadata_bits,
       "E_trace_bits_removed":legacy_bits - (blocks_seen*512),
       "interpretation":"schedule tail is an invertible recoding of each block, not a reduction of message information"})
````

The SHA observer returns because its sidecar contains message-derived information, so it is not a digest-only inverse.
