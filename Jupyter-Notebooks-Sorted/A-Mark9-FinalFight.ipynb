```python
# ============================================
# PAIRWISE SUBTRACTION ENCODER
# Input: any integer → Output: all digit sequences that encode it
# ============================================

TARGET = 14  # ← SET YOUR VALUE HERE

def find_encoding_sequences(target_int):
    """
    Recurse up the tree: find all digit sequences where adjacent 
    absolute differences equal the digits of target_int.
    """
    target_digits = [int(d) for d in str(target_int)]
    n_gaps = len(target_digits)
    solutions = []
    
    def recurse_up(current_seq, gap_index):
        # Base: filled all gaps
        if gap_index >= n_gaps:
            solutions.append(tuple(current_seq))
            return
        
        needed_gap = target_digits[gap_index]
        last_digit = current_seq[-1]
        
        # Two branches: last_digit + gap, or last_digit - gap
        # Must stay 0-9 for single-digit constraint
        for direction in [1, -1]:
            next_digit = last_digit + (needed_gap * direction)
            if 0 <= next_digit <= 9:
                recurse_up(current_seq + [next_digit], gap_index + 1)
    
    # Seed from any valid starting digit 0-9
    for seed in range(10):
        recurse_up([seed], 0)
    
    return solutions

# EXECUTE
sequences = find_encoding_sequences(TARGET)

print(f"TARGET: {TARGET}")
print(f"Digits to encode: {list(str(TARGET))}")
print(f"Gaps needed: {len(str(TARGET))}")
print(f"\nFound {len(sequences)} valid encoding sequences:\n")

for seq in sequences:
    # Verify
    gaps = [abs(seq[i+1] - seq[i]) for i in range(len(seq)-1)]
    print(f"  {seq} → {''.join(str(g) for g in gaps)}")

print(f"\n{'='*50}")
print(f"TOTAL: {len(sequences)} sequences")
print(f"{'='*50}")
```

```python
# ============================================================
# REVERSE ABS-DIFF EXPANDER
# seed = collapsed data row
# target_len = desired row length
# expands upward until row length == target_len
# prints full stack for every valid solution
# ============================================================

SEED = "14159265358"   # or "229843"
TARGET_LEN = 16

def expand_one_row(child_row):
    """
    Given child_row of length n,
    find all parent rows of length n+1 such that:
        abs(parent[i+1] - parent[i]) == child_row[i]
    Digits are restricted to 0..9.
    """
    out = []

    def build(parent):
        i = len(parent) - 1
        if i == len(child_row):
            out.append(parent[:])
            return

        gap = child_row[i]
        last = parent[-1]

        a = last + gap
        b = last - gap

        if 0 <= a <= 9:
            build(parent + [a])

        if gap != 0 and 0 <= b <= 9:
            build(parent + [b])

    for start in range(10):
        build([start])

    return out


def expand_to_length(seed_digits, target_len):
    """
    Returns all stacks from seed row up to target_len.
    Each stack is [row_len_seed, row_len_seed+1, ..., row_len_target]
    """
    current_stacks = [[seed_digits[:]]]

    while len(current_stacks[0][-1]) < target_len:
        next_stacks = []
        for stack in current_stacks:
            child = stack[-1]
            parents = expand_one_row(child)
            for p in parents:
                next_stacks.append(stack + [p])
        current_stacks = next_stacks
        if not current_stacks:
            break

    return current_stacks


def row_to_str(row):
    return "".join(str(x) for x in row)


def verify_stack(stack):
    for i in range(len(stack) - 1):
        parent = stack[i + 1]
        child = stack[i]
        derived = [abs(parent[j+1] - parent[j]) for j in range(len(parent) - 1)]
        if derived != child:
            return False
    return True


seed_digits = [int(c) for c in str(SEED)]

if TARGET_LEN < len(seed_digits):
    raise ValueError("TARGET_LEN must be >= len(SEED)")

stacks = expand_to_length(seed_digits, TARGET_LEN)

print(f"SEED={SEED}")
print(f"TARGET_LEN={TARGET_LEN}")
print(f"SOLUTIONS={len(stacks)}")

for s_idx, stack in enumerate(stacks, 1):
    print("=" * 80)
    print(f"SOLUTION {s_idx}")
    print("=" * 80)
    for level, row in enumerate(stack):
        print(f"LEN={len(row):02d}  {row_to_str(row)}")
    print(f"VERIFY={verify_stack(stack)}")
```
