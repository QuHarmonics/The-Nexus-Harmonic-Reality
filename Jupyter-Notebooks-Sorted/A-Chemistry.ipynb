$$

$$
G_k = \{ Z \mid 2^{k-1} \le Z \le 2^k - 1 \}.

```python
import math
from rdkit.Chem import GetPeriodicTable
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.express as px

def build_dataframe() -> pd.DataFrame:
    pt = GetPeriodicTable()
    rows = []
    for z in range(1, 119):
        rows.append({
            "Atomic Number": z,
            "Symbol": pt.GetElementSymbol(z),
            "Element": pt.GetElementName(z),
            "Atomic Mass": float(pt.GetAtomicWeight(z)),
        })

    df = pd.DataFrame(rows)

    # Core binary-length grouping
    df["Binary Length"] = df["Atomic Number"].apply(lambda z: int(math.floor(math.log2(z))) + 1)
    df["Binary Group Start"] = df["Binary Length"].apply(lambda k: 2 ** (k - 1))
    df["Binary Group End"] = df["Binary Length"].apply(lambda k: 2 ** k - 1)

    # First difference in atomic mass
    df["Mass Delta"] = df["Atomic Mass"].diff()

    df["Group Label"] = df.apply(
        lambda r: f"{r['Binary Length']}-bit [{int(r['Binary Group Start'])}-{int(r['Binary Group End'])}]",
        axis=1
    )
    return df

def build_figure(df: pd.DataFrame) -> go.Figure:
    group_stats = (
        df.groupby(["Binary Length", "Binary Group Start", "Binary Group End"])
        .agg(
            Count=("Atomic Number", "size"),
            MinZ=("Atomic Number", "min"),
            MaxZ=("Atomic Number", "max"),
            MeanMass=("Atomic Mass", "mean"),
            MeanDelta=("Mass Delta", "mean"),
        )
        .reset_index()
    )

    colors = px.colors.qualitative.Bold

    fig = make_subplots(
        rows=3,
        cols=1,
        shared_xaxes=True,
        vertical_spacing=0.06,
        row_heights=[0.48, 0.28, 0.24],
        subplot_titles=(
            "Atomic Mass by Atomic Number, colored by binary-length group",
            "Mass delta between consecutive elements",
            "Count of elements in each binary-length container",
        ),
    )

    # Shade binary-length containers behind first two panels
    for idx, row in group_stats.iterrows():
        start = int(row["Binary Group Start"])
        end = int(row["Binary Group End"])
        shown_end = min(end, 118)
        color = colors[idx % len(colors)]

        for r in [1, 2]:
            fig.add_vrect(
                x0=start - 0.5,
                x1=shown_end + 0.5,
                fillcolor=color,
                opacity=0.07,
                line_width=0,
                row=r,
                col=1,
            )

        fig.add_annotation(
            x=(start + shown_end) / 2,
            y=1.05,
            xref="x",
            yref="paper",
            text=f"{int(row['Binary Length'])}-bit<br>{start}–{end}",
            showarrow=False,
            font=dict(size=10, color=color),
        )

    # Atomic mass traces by binary-length group
    for idx, (k, grp) in enumerate(df.groupby("Binary Length")):
        color = colors[idx % len(colors)]
        hover = [
            f"Z={z}<br>{sym} ({name})<br>Mass={mass:.6f}<br>{k}-bit group"
            for z, sym, name, mass in zip(
                grp["Atomic Number"], grp["Symbol"], grp["Element"], grp["Atomic Mass"]
            )
        ]

        fig.add_trace(
            go.Scatter(
                x=grp["Atomic Number"],
                y=grp["Atomic Mass"],
                mode="lines+markers",
                name=f"{k}-bit group",
                legendgroup=f"group{k}",
                marker=dict(size=7, color=color),
                line=dict(width=2, color=color),
                text=hover,
                hoverinfo="text",
            ),
            row=1,
            col=1,
        )

    # Mass delta bar plot
    fig.add_trace(
        go.Bar(
            x=df["Atomic Number"][1:],
            y=df["Mass Delta"][1:],
            marker_color="#636EFA",
            name="Mass Δ",
            showlegend=False,
            hovertemplate="Z=%{x}<br>Δmass=%{y:.6f}<extra></extra>",
        ),
        row=2,
        col=1,
    )

    # Counts per binary container
    fig.add_trace(
        go.Bar(
            x=[str(int(k)) for k in group_stats["Binary Length"]],
            y=group_stats["Count"],
            marker_color=[colors[i % len(colors)] for i in range(len(group_stats))],
            text=[f"{int(c)}" for c in group_stats["Count"]],
            textposition="outside",
            name="Count",
            showlegend=False,
            hovertemplate="%{x}-bit group<br>count=%{y}<extra></extra>",
        ),
        row=3,
        col=1,
    )

    # Mark power-of-two boundaries
    for n in range(1, 8):
        x = 2 ** (n - 1)
        for r in [1, 2]:
            fig.add_vline(
                x=x - 0.5,
                line_dash="dot",
                line_color="gray",
                opacity=0.45,
                row=r,
                col=1,
            )

    fig.update_layout(
        height=980,
        width=1200,
        title_text="Periodic Table Grouping by Binary Length of Atomic Number",
        template="plotly_white",
        legend_title_text="Binary Length",
        bargap=0.12,
    )

    fig.update_xaxes(title_text="Atomic Number Z", row=2, col=1)
    fig.update_xaxes(title_text="Binary Length (bits)", row=3, col=1, type="category")
    fig.update_yaxes(title_text="Atomic Mass", row=1, col=1)
    fig.update_yaxes(title_text="Δ Atomic Mass", row=2, col=1)
    fig.update_yaxes(title_text="Element Count", row=3, col=1)

    return fig

# Run
df = build_dataframe()
display(df.head(20))

fig = build_figure(df)
fig.show()
```
